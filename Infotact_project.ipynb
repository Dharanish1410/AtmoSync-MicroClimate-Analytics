```python
datasets = [products, components, warranty, resale]

for df in datasets:
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

print("Column names cleaned!")
```

```python
for df in datasets:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

print("Extra spaces removed!")
```

```python
for name, df in zip(
    ["Products", "Components", "Warranty", "Resale"],
    datasets
):
    print(name)
    print(df.isnull().sum())
```

```python
for df in datasets:
    for col in df.select_dtypes(include="number").columns:
        df[col] = df[col].fillna(df[col].median())

    for col in df.select_dtypes(include="object").columns:
        if df[col].isnull().any():
            mode_value = df[col].mode()
            if not mode_value.empty:
                df[col] = df[col].fillna(mode_value[0])

print("Missing values handled!")
```

```python
warranty["failure_date"] = pd.to_datetime(
    warranty["failure_date"], errors="coerce"
)

resale["listing_date"] = pd.to_datetime(
    resale["listing_date"], errors="coerce"
)

print("Date columns converted successfully!")
```

```python
products["original_price_inr"] = pd.to_numeric(
    products["original_price_inr"], errors="coerce"
)

components["component_cost_inr"] = pd.to_numeric(
    components["component_cost_inr"], errors="coerce"
)

warranty["repair_cost_inr"] = pd.to_numeric(
    warranty["repair_cost_inr"], errors="coerce"
)

resale["resale_price_inr"] = pd.to_numeric(
    resale["resale_price_inr"], errors="coerce"
)

print("Numerical columns converted!")
```

```python
# Standardize text values

for df in datasets:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

# Convert recoverable column to uppercase string
components["recoverable"] = (
    components["recoverable"]
    .astype("string")
    .str.strip()
    .str.upper()
)

print("Text values standardized successfully!")
```

```python
print("Invalid Product Prices:")
display(products[products["original_price_inr"] <= 0])

print("Invalid Component Costs:")
display(components[components["component_cost_inr"] < 0])

print("Invalid Repair Costs:")
display(warranty[warranty["repair_cost_inr"] < 0])

print("Invalid Resale Prices:")
display(resale[resale["resale_price_inr"] <= 0])
```

```python
print("Unmatched Component Product IDs:")
display(
    components[~components["product_id"].isin(products["product_id"])]
)

print("Unmatched Warranty Product IDs:")
display(
    warranty[~warranty["product_id"].isin(products["product_id"])]
)

print("Unmatched Resale Product IDs:")
display(
    resale[~resale["product_id"].isin(products["product_id"])]
)
```
